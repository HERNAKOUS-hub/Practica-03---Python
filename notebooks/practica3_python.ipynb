## Ejercicio 1 — Catálogo de productos como lista
Una tienda gestiona su catálogo en una lista plana que alterna nombre de producto y precio. Trabaja con esta estructura sin convertirla todavía a diccionario.

```python
catalogo = [
    "Auriculares BT", 59.90,
    "Smartwatch S2", 149.00,
    "Tablet 10", 219.00,
    "Cargador USB-C", 18.50,
    "Cafetera Expres", 189.00,
    "Robot Aspirador", 279.00,
]
```

**Se pide:**

1. Imprime el número total de elementos de la lista y deduce cuántos productos hay.
2. Usando *slicing con paso*, crea `nombres` con todos los nombres y `precios` con todos los precios.
3. Obtén el precio del `"Robot Aspirador"` **sin escribir el índice a mano**: localiza su posición con `.index()` y usa esa posición para llegar al precio.
4. Crea `ultimos_tres_nombres` con los tres últimos productos usando índices negativos.
5. Calcula e imprime el precio medio del catálogo redondeado a dos decimales, usando únicamente `sum()`, `len()` y `round()`.

````python
catalogo = [
    "Auriculares BT", 59.90,
    "Smartwatch S2", 149.00,
    "Tablet 10", 219.00,
    "Cargador USB-C", 18.50,
    "Cafetera Expres", 189.00,
    "Robot Aspirador", 279.00,
]

# 1. Número total de elementos y deducción
total_elementos = len(catalogo)
cantidad_productos = total_elementos // 2
print(f"Total de elementos en la lista: {total_elementos}")
print(f"Dado que cada producto tiene 2 elementos (nombre y precio), hay {cantidad_productos} productos.\n")

# 2. Slicing con paso para nombres y precios
nombres = catalogo[0::2]  # Empieza en el índice 0, salta de 2 en 2
precios = catalogo[1::2]  # Empieza en el índice 1, salta de 2 en 2
print(f"Nombres (Total elementos: {len(nombres)}): {nombres}")
print(f"Precios (Suma total: {sum(precios):.2f}): {precios}\n")

# 3. Obtener precio del "Robot Aspirador"
indice_robot = catalogo.index("Robot Aspirador")
precio_robot = catalogo[indice_robot + 1] # El precio está en la posición inmediatamente siguiente
print(f"El precio del Robot Aspirador es: {precio_robot}\n")

# 4. Últimos tres nombres usando índices negativos
# Aplicado directamente sobre la lista 'nombres' generada previamente
ultimos_tres_nombres = nombres[-3:]
# (Alternativa sobre el catálogo original: catalogo[-6::2])
print(f"Los últimos tres nombres son: {ultimos_tres_nombres}\n")

# 5. Precio medio del catálogo
precio_medio = round(sum(precios) / len(precios), 2)
print(f"El precio medio del catálogo es: {precio_medio}")
````

## Ejercicio 2 — Inventario multialmacén con listas anidadas
El inventario de tres almacenes se representa como una lista de listas. Cada sublista sigue el formato `[codigo_almacen, producto, unidades, coste_unitario]`.

```python
inventario = [
    ["ALM-NORTE", "Robot Aspirador", 34, 201.50],
    ["ALM-NORTE", "Monitor 27", 58, 142.00],
    ["ALM-SUR",   "Silla Ergonomica", 12, 178.90],

    ["ALM-ESTE",  "Cafetera Expres", 41, 131.20],
    ["ALM-ESTE",  "Monitor 27", 25, 142.00],
]
```

**Se pide:**

1. Imprime el nombre del producto de la **cuarta** fila usando doble indexación.
2. Crea una nueva lista `valor_stock` donde cada elemento sea `[producto, unidades * coste_unitario]` redondeado a dos decimales. Resuélvelo con un bucle `for` sobre `inventario`.
3. Calcula el **valor inmovilizado total** del inventario.
4. Extrae, mediante *slicing*, una sublista con solo las filas del `ALM-SUR` (usa el hecho de que están contiguas) y calcula su valor.
5. Explica en una celda Markdown de **dos frases** qué problema aparece si dos almacenes dejan de estar contiguos y por qué el *slicing* no es una estrategia robusta de filtrado.

**Comprobación:** el valor inmovilizado total debe ser 27 573,50 € y el del `ALM-SUR`, 3 557,30 €.

````python
inventario = [
    ["ALM-NORTE", "Robot Aspirador", 34, 201.50],
    ["ALM-NORTE", "Monitor 27", 58, 142.00],
    ["ALM-SUR",   "Silla Ergonomica", 12, 178.90],
    ["ALM-SUR",   "Robot Aspirador", 7, 201.50],
    ["ALM-ESTE",  "Cafetera Expres", 41, 131.20],
    ["ALM-ESTE",  "Monitor 27", 25, 142.00],
]

# 1. Nombre del producto de la cuarta fila (índice 3, columna 1)
producto_cuarta_fila = inventario[3][1]
print(f"Producto de la cuarta fila: {producto_cuarta_fila}\n")

# 2. Crear la lista valor_stock con un bucle for
valor_stock = []
for fila in inventario:
    producto = fila[1]
    valor = round(fila[2] * fila[3], 2)
    valor_stock.append([producto, valor])

print("Valor del stock por producto:")
for item in valor_stock:
    print(item)

# 3. Calcular el valor inmovilizado total
valor_total = sum(item[1] for item in valor_stock)
print(f"\nValor inmovilizado total: {valor_total:.2f} €")

# 4. Slicing de ALM-SUR (índices 2 y 3) y su valor
inventario_sur = inventario[2:4]
valor_sur = round(sum(fila[2] * fila[3] for fila in inventario_sur), 2)
print(f"Valor inmovilizado ALM-SUR: {valor_sur:.2f} €")
````

## Ejercicio 4 — Cola de pedidos pendientes

La cola de pedidos de un almacén se gestiona como lista ordenada por prioridad.

```python
cola = ["PED-10021", "PED-10007", "PED-10044", "PED-10012", "PED-10033", "PED-10008"]
```

**Se pide:**

1. Invierte la cola usando *slicing* con paso negativo (sin `.reverse()`) y guárdala en `cola_invertida`.
2. Ordena `cola` alfabéticamente con `sorted()` **sin modificar la original** y comprueba que la original sigue intacta.
3. Ordena ahora la lista **en sitio** con `.sort(reverse=True)`.
4. Atiende los dos primeros pedidos: extráelos con `.pop(0)` y guárdalos en `atendidos`.
5. Inserta un pedido urgente `"PED-99999"` al principio de la cola.
6. Imprime un informe final con el número de pedidos atendidos, los pendientes y el porcentaje de avance redondeado a un decimal.

````python
cola = ["PED-10021", "PED-10007", "PED-10044", "PED-10012", "PED-10033", "PED-10008"]

# 1. Invertir la cola usando slicing con paso negativo
cola_invertida = cola[::-1]
print(f"1. Cola invertida: {cola_invertida}\n")

# 2. Ordenar con sorted() comprobando que la original no cambia
cola_alfabetica = sorted(cola)
print(f"2. Ordenada con sorted(): {cola_alfabetica}")
print(f"   Comprobación original: {cola}\n")

# 3. Ordenar en sitio con .sort(reverse=True)
cola.sort(reverse=True)
print(f"3. Ordenada en sitio (inversa): {cola}\n")

# 4. Extraer los dos primeros pedidos
atendidos = []
atendidos.append(cola.pop(0))
atendidos.append(cola.pop(0))
print(f"4. Pedidos atendidos: {atendidos}")
print(f"   Cola restante: {cola}\n")

# 5. Insertar pedido urgente al principio
cola.insert(0, "PED-99999")
print(f"5. Cola con pedido urgente: {cola}\n")

# 6. Informe final
num_atendidos = len(atendidos)
num_pendientes = len(cola)
total_pedidos = num_atendidos + num_pendientes
porcentaje_avance = round((num_atendidos / total_pedidos) * 100, 1)

print("--- INFORME FINAL ---")
print(f"Pedidos atendidos:  {num_atendidos}")
print(f"Pedidos pendientes: {num_pendientes}")
print(f"Avance de la cola:  {porcentaje_avance}%")
````

## Ejercicio 5 — Funciones propias para el cálculo de importes

**Se pide:**

1. Define la función `calcular_importe(unidades, precio_unitario, descuento_pct=0)` que devuelva el importe neto redondeado a dos decimales. El parámetro `descuento_pct` debe tener valor por defecto.
2. Define `resumen_pedido(unidades, precio_unitario, descuento_pct=0)` que devuelva **una tupla de tres elementos**: `(importe_bruto, ahorro, importe_neto)`.
3. Llama a `resumen_pedido` usando **argumentos por palabra clave** y desempaqueta el resultado en tres variables.
4. Documenta ambas funciones con *docstring* y muestra la ayuda con `help(calcular_importe)`.
5. Usa `max()`, `min()`, `sorted()` y `len()` sobre la lista `importes` que obtengas al aplicar `calcular_importe` a estos cinco pedidos:
    
    ```python
    pedidos = [
        (3, 59.90, 10),
        (1, 279.00, 0),
        (5, 18.50, 20),
        (2, 149.00, 5),
        (4, 89.90, 15),
    ]
    ```
    
    **Comprobación:** `calcular_importe(3, 59.90, 10)` debe devolver `161.73`.

````python
def calcular_importe(unidades, precio_unitario, descuento_pct=0):
    """
    Calcula el importe neto de un pedido aplicando un descuento porcentual.
    
    Parámetros:
    unidades (int): Número de artículos.
    precio_unitario (float): Precio por unidad.
    descuento_pct (int/float, opcional): Porcentaje de descuento (por defecto 0).
    
    Devuelve:
    float: Importe neto final redondeado a 2 decimales.
    """
    importe_bruto = unidades * precio_unitario
    importe_neto = importe_bruto * (1 - descuento_pct / 100)
    return round(importe_neto, 2)

def resumen_pedido(unidades, precio_unitario, descuento_pct=0):
    """
    Genera un desglose detallado del pedido.
    
    Devuelve:
    tuple: (importe_bruto, ahorro, importe_neto)
    """
    importe_bruto = unidades * precio_unitario
    ahorro = importe_bruto * (descuento_pct / 100)
    importe_neto = importe_bruto - ahorro
    return round(importe_bruto, 2), round(ahorro, 2), round(importe_neto, 2)

# Llamada con argumentos por palabra clave y desempaquetado
bruto, ahorro, neto = resumen_pedido(unidades=3, precio_unitario=59.90, descuento_pct=10)
print(f"Desglose pedido -> Bruto: {bruto} €, Ahorro: {ahorro} €, Neto: {neto} €\n")

# Mostrar la ayuda (docstring) de la función
help(calcular_importe)

# Cálculos sobre la lista de pedidos
pedidos = [
    (3, 59.90, 10),
    (1, 279.00, 0),
    (5, 18.50, 20),
    (2, 149.00, 5),
    (4, 89.90, 15),
]

# Crear la lista de importes aplicando la función
importes = [calcular_importe(u, p, d) for u, p, d in pedidos]
print(f"\nImportes calculados: {importes}")

# Uso de funciones integradas sobre la lista
print(f"Importe máximo (max) : {max(importes)} €")
print(f"Importe mínimo (min) : {min(importes)} €")
print(f"Lista ordenada (sorted): {sorted(importes)}")
print(f"Total de importes (len)  : {len(importes)}")
````

## Ejercicio 6 — Normalización de códigos SKU con métodos

El equipo de logística recibe códigos de producto en formatos inconsistentes.

```python
skus_crudos = [
    "  elc-0012-es ", "HOG-0045-ES", "dep-0003-pt  ",
    "ofi-0021-es", "  ELC-0012-ES", "hog-0099-fr ",
]
```

**Se pide:**

1. Crea `skus_limpios` aplicando a cada código: eliminación de espacios sobrantes (`.strip()`) y conversión a mayúsculas (`.upper()`).
2. Cuenta cuántas veces aparece `"ELC-0012-ES"` en la lista limpia con `.count()`.
3. Obtén la posición de la primera aparición de `"OFI-0021-ES"` con `.index()`.
4. Para cada SKU limpio, usa `.split("-")` para separar familia, número y país. Construye una lista de listas con esas tres partes.
5. Crea `skus_es` con solo los SKU cuyo país sea `"ES"` y sustituye en todos ellos el guion por un punto con `.replace()`.
6. Genera una cadena única separada por `" | "` usando `.join()` y muéstrala.

> **Pista:** `.strip()`, `.upper()` y `.replace()` **devuelven** una nueva cadena; no modifican la original. `.append()` y `.sort()` sobre listas, en cambio, actúan en sitio y devuelven `None`.
>

````python
skus_crudos = [
    "  elc-0012-es ", "HOG-0045-ES", "dep-0003-pt  ",
    "ofi-0021-es", "  ELC-0012-ES", "hog-0099-fr ",
]

# 1. Crear skus_limpios quitando espacios y pasando a mayúsculas
skus_limpios = [sku.strip().upper() for sku in skus_crudos]
print(f"1. SKUs limpios: {skus_limpios}\n")

# 2. Contar cuántas veces aparece "ELC-0012-ES"
conteo = skus_limpios.count("ELC-0012-ES")
print(f"2. Apariciones de 'ELC-0012-ES': {conteo}\n")

# 3. Obtener la posición de "OFI-0021-ES"
posicion = skus_limpios.index("OFI-0021-ES")
print(f"3. Índice de 'OFI-0021-ES': {posicion}\n")

# 4. Separar familia, número y país en una lista de listas
partes_sku = [sku.split("-") for sku in skus_limpios]
print(f"4. Partes separadas: {partes_sku}\n")

# 5. Filtrar por 'ES' y cambiar guiones por puntos
skus_es = [sku.replace("-", ".") for sku in skus_limpios if sku.endswith("-ES")]
print(f"5. SKUs de España (con puntos): {skus_es}\n")

# 6. Unir con ' | '
cadena_final = " | ".join(skus_es)
print(f"6. Cadena unida: {cadena_final}")
````

    ```python
    seleccion = (np_precios > 100) & (np_unidades >= 2)

    ```
    

**Comprobación:** hay 4 productos con precio unitario superior a 100 € y la suma de sus importes es 1 254,00 €.

> **Pista:** con NumPy no se usan `and` ni `or`, sino `&` (y) y `|` (o). Sin los paréntesis el resultado es un error, porque `&` se evalúa antes que `>`.
>

````python
import numpy as np

# Recreamos los arrays del ejercicio anterior para que funcione la celda
np_unidades = np.array([3, 1, 5, 2, 4, 6, 2, 8, 3, 1])
np_precios = np.array([59.90, 279.00, 18.50, 149.00, 89.90, 27.50, 219.00, 14.90, 64.00, 239.00])
importes = np_unidades * np_precios

# 1. Crear máscara booleana
caros = np_precios > 100
print(f"1. Máscara 'caros':\n{caros}\n")

# 2. Extraer precios superiores a 100
precios_caros = np_precios[caros]
print(f"2. Precios > 100 €:\n{precios_caros}\n")

# 3. Contar productos caros
cantidad_caros = caros.sum()
print(f"3. Cantidad de productos caros: {cantidad_caros}\n")

# 4. Importes de esas líneas y su suma
importes_caros = importes[caros]
suma_importes = importes_caros.sum()
print(f"4. Importes de líneas caras:\n{importes_caros}")
print(f"   Suma de importes: {suma_importes:.2f} €\n")

# 5. Posiciones con np.where
posiciones = np.where(caros)
print(f"5. Índices de los productos caros:\n{posiciones[0]}\n")

# 6. Segunda máscara y combinación
varias = np_unidades >= 2
seleccion = (np_precios > 100) & (np_unidades >= 2)
print(f"6. Importes con precio >100 y unidades >=2:\n{importes[seleccion]}")
````

#### Punto 5

Sumar booleanos devuelve un número porque Python evalúa internamente True como el número entero 1 y False como 0; al sumarlos, simplemente estás contando cuántas veces se cumple la condición.

## Ejercicio 10 — Matriz de ventas por trimestre y región

```python
import numpy as np

# Filas: T1, T2, T3  |  Columnas: Norte, Sur, Este, Oeste, Centro
ventas_2d = np.array([
    [15200.50, 11800.25, 12950.00,  9870.40, 13400.10],
    [16840.75, 12310.60, 13480.90, 11020.35, 14100.80],
    [12535.72, 10663.06, 11221.59, 12673.08, 12334.38],
])
```

**Se pide:**

1. Imprime `shape`, `ndim` y `size` del array y explica qué representa cada uno en este contexto de negocio.
2. Extrae la fila del segundo trimestre y la columna de la región Centro.
3. Obtén el valor de T3 en la región Este mediante indexación 2D.
4. Extrae con *slicing* la submatriz de T1 y T2 para las tres primeras regiones.
5. Calcula el total por trimestre (`axis=1`) y el total por región (`axis=0`).
6. Aplica una previsión de crecimiento del 6 % a T3 sumando el resultado como una **cuarta fila** con `np.vstack()`.
7. Identifica la región con mayor facturación acumulada usando `np.argmax()` sobre los totales por región, y traduce el índice a nombre con una lista de etiquetas.

````python
import numpy as np

# Filas: T1, T2, T3  |  Columnas: Norte, Sur, Este, Oeste, Centro
ventas_2d = np.array([
    [15200.50, 11800.25, 12950.00,  9870.40, 13400.10],
    [16840.75, 12310.60, 13480.90, 11020.35, 14100.80],
    [12535.72, 10663.06, 11221.59, 12673.08, 12334.38],
])

# 1. Propiedades del array
print(f"1. Propiedades de la matriz:")
print(f"   shape: {ventas_2d.shape}")
print(f"   ndim:  {ventas_2d.ndim}")
print(f"   size:  {ventas_2d.size}\n")

# 2. Extracción de fila T2 y columna Centro
fila_t2 = ventas_2d[1, :]      # Índice 1 (segunda fila)
columna_centro = ventas_2d[:, 4] # Índice 4 (quinta columna)
print(f"2. Ventas T2 (todas las regiones): {fila_t2}")
print(f"   Ventas Centro (todos los trimestres): {columna_centro}\n")

# 3. Valor de T3 en Este
valor_t3_este = ventas_2d[2, 2] # Fila 2, Columna 2
print(f"3. Valor en T3, región Este: {valor_t3_este} €\n")

# 4. Slicing submatriz (T1 y T2 para Norte, Sur, Este)
submatriz = ventas_2d[0:2, 0:3]
print(f"4. Submatriz T1-T2 y 3 primeras regiones:\n{submatriz}\n")

# 5. Totales por eje
totales_trimestre = ventas_2d.sum(axis=1) # Sumar a lo largo de las columnas (horizontal)
totales_region = ventas_2d.sum(axis=0)    # Sumar a lo largo de las filas (vertical)
print(f"5. Total por trimestre: {totales_trimestre}")
print(f"   Total por región: {totales_region}\n")

# 6. Previsión T4 (+6% sobre T3) y añadido con vstack
t4_prevision = ventas_2d[2, :] * 1.06
ventas_ampliadas = np.vstack((ventas_2d, t4_prevision))
print(f"6. Matriz ampliada con T4:\n{ventas_ampliadas}\n")

# 7. Región con mayor facturación
regiones = ["Norte", "Sur", "Este", "Oeste", "Centro"]
indice_max = np.argmax(totales_region)
mejor_region = regiones[indice_max]
print(f"7. Región con mayor facturación: {mejor_region} (Total: {totales_region[indice_max]:.2f} €)")
````

#### Apartado 1

shape (3, 5): Representa la estructura tabular del negocio: 3 trimestres analizados (filas) a través de 5 regiones geográficas (columnas).

ndim 2: Confirma que estamos trabajando con una estructura bidimensional plana (tabla cruzada de doble entrada).

size 15: Es el volumen total de observaciones; 15 datos individuales de facturación recogidos en la matriz.
